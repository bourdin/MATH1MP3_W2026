# Chapter 6: functions

Functions are an essential step towards write *reusable code*. If the same block of code is reused repeatedly, a function is a way to repeat it multiple code without rewriting it each time. We have encountered many functions already. For instance, when we execute the command `print("Hello world")`, we call a python function (`print`) which does whatever is needed so that he message "hello world" appears in our notebook. In this context, we say that we *call* the function `print` with argument `"hello world"`. 

## 6.1 Defining and calling a function:

A function is defined using the keyword `def` for instance:

```python
def sayHello():
    print("Hello!")
```

The *body* of the function (*i.e.* everything that it should do) is denoted through indentation.
Note the code *inside* the function is not executed here. For that, we need to *call* the function:

```python
sayHello()
```

## 6.2 Arguments
The example above is a bit limited. A better use of function is to perform actions that depend on some parameters. This is typically done by *passing arguments* to a function.

Consider the example below:

```python
def printRectangleArea(width, height):
    print(f"the area of a {width} x {height} rectangle is {width * height}")
```

We can now execute the function multiple times for different values of `width` and `height`: 

```python
printRectangleArea(1, 2)
printRectangleArea(1, 5)
printRectangleArea(2, 3)
printRectangleArea(10, 2)
```

Here the variables `width` and `height` behave like the independent variable of a mathematical function. They don't have a real meaning *per se*. They only exist inside the function `printRectangleArea` where they take the value that was *passed* when the function was called. We say that the *scope* of these variables is limited to the function

```python
printRectangleArea(1, 2)
print('height is ', height)
```

Python is a bit special amongst programming languages in that it does not care about the type of the arguments of a function (this is called a polymorphic language).
This can be nice in some cases, for instance:

```python
def printSum(a,b):
    print(f'{a} + {b} = {a+b}')

printSum(1,2)
printSum(1.23, 3.45)
printSum("hello", " world")
```

```python
printSum(1, "a")
```

It is however considered good practice to indicate the type of variables that a function expects, so that somebody reading code understands the intent of the author. In the previous case, if what we intended was only to add integers, a better definition of the function would be:  

```python
def printSum2(a: int, b: int):
    print(f'{a} + {b} = {a+b}')

printSum2(1,2)
```

Note that this does not affect the ability of our function to take other kind of arguments. It just makes it easier to read for others (like the professors and TA who will grade your work)...

```python
printSum2(1.23, 3.45)
printSum2("hello", " world")
```

## 6.3 Return values

Functions are often used to 'compute' a quantity that we may want to reuse instead of just doing something. Back to the addition example, maybe we want to reuse the result of this function. In order to do this, we need to tell the function to *return* a result:

```python
def addInt(a: int, b: int) -> int:
    sum = a + b
    return sum
```

```python
c = addInt(1, 2)
print(c)
```

We could also have done 

```python
print(addInt(1, 2))
```

In your vector calculus class, you will study functions that given a scalar, return a vector. Similarly, python functions can return results of any type, or even several results. See for instance

```python
def sumAndProduct(a: float, b: float) -> float:
    return a + b, a * b
```

```python
sumAndProduct(2,3)
```

```python
s,p = sumAndProduct(2,3)
print(s, p)
```

```python
(s,p) = sumAndProduct(2,3)
print(s, p)
```

## 6.4 adding documentation
Ideally, the name of your function should give a clear indication of what it does.
Technically, there is nothing wrong with the function below:

```python
def sumFloat(a, b):
    return a*b
```

But it will lead to code that is impossible to understand. 

Sometimes, the name of the function if not sufficient to fully describe what it does. It is good practice to add a description of what a function does, how it does it etc.
Python encourages this behaviour by automatically turning this description in to help. Note the triple quotes. These are used to allow text that spans multiple lines.

```python
def sumFloat(a: float, b: float) -> float:
    '''
    Returns the sum of two numbers. It is a bit of a silly function but
    it does a decent job at illustrating how python encourage documenting functions
    '''
    return a + b
```

```python
help(sumFloat)
```
