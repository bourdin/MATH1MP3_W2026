### Root finding Part I - Bisection method

Given a function $f:\mathbb{R}\rightarrow\mathbb{R}$ we wanted to devise a computer assisted way of answering the question: If $f$ is known to have a root on the interval $[a,b]$ (possibly a large interval), where exactly does this root live? One can observe that without making any restrictions on the local behaviour of $f$, this may be a very difficult question. For instance, consider the following function $g(x)$,
$$g(x)=\begin{cases} 0 & \text{ if } x=\pi\\
2 & \text{ else}
\end{cases}$$

As $g:\mathbb{R}\rightarrow\mathbb{R}$ is not continous, then the local behaviour of $g$ near a root does not lead us nearer to the root. 

Given the assumption of continuity an important theorem from caclulus becomes a great help.

**Theorem (IVT):** Let $f:\mathbb{R}\rightarrow\mathbb{R}$ be continuous on $[a,b]\subseteq\mathbb{R}$. If there exists $a\leq c\leq d\leq b$, such that $f(c)\leq \alpha \leq f(d)$, then there exists $\beta\in [c,d]$, such that $f(\beta)=\alpha$. 

Namely, we can see that if $\alpha=0$, then this allows us to shrink the interval $[a,b]$ to a smaller interval $[c,d]$ that must contain the root. 

This observation allowed us to devise the bisection algorithm for root finding. 

```python
# we need this import statement to type annotate parameters
# of function type.
from collections.abc import Callable

def bisection(f : Callable[[float],float],a : float,b : float,tol : float) -> float | tuple[float,float] :
    '''
    An implementation of the bisection method for root finding.
    Takes as input a function `f`, an interval `[a,b]`, and a tolerance
    `tol`. Proceeds as follows:
    0. Let c = (a+b)/2 (the midpoint)
    1.  If f(a) = 0 return a
        If f(b) = 0 return b
        If f(c) = 0 return c
    2. Check if abs(b-a) < tol:
        If yes, return `(a,b)`.
        Otherwise...
    3. If f(a)*f(c)<0:
        b = c
        If f(c)*f(b)<0:
        a = c
    4. Repeat!
    '''
    if f(a)*f(b)< 0:
        while abs(a - b) >= tol:
            c = (a + b)/2 #midpoint of the interval
            if f(a) == 0.0 :
                return f"Found a root at x = {a}."
            elif f(b) == 0.0:
                return f"Found a root a x = {b}."
            elif f(c) == 0.0:
                return f"Found a root a x = {c}."
            elif f(a)*f(c) < 0:
                b = c
            else:
                a = c
        return f'Last interval checked ({a},{b})'
    else:
        return f" I don't know ask Newton"
```

```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np

def Newton(f,fprime,a,eps, nMax = 100):
    numIter = 0
    err = eps + 1 # initializing the error
    # could also start with abs(f(a))
    while numIter < nMax and err > eps:
        numIter += 1
        c = a-f(a) / fprime(a)
        err = abs(f(c))
        a = c
        # this print call is a side effect of the function Newton
        print(f"After {numIter} iteration(s) obtained x = {c} with error: {err}")
    return c

def f(x):
    import numpy as np
    return np.cos(x)
def fprime(x):
    import numpy as np
    return -np.sin(x)

    

a = -np.pi/4
b = np.pi
eps = 1.e-3
print(Newton(f, fprime, a, eps, 10))
```

```python
X = np.linspace(a-(b-a)/2, b + (b-1)/2,101)
fig, ax = plt.subplots(figsize = (8,6))
ax.plot(X,f(X))
ax.axis((a-(b-a)/2, b + (b-1)/2,-1,1))

def NewtonWithPlot(f, fprime, a, b, eps, nMax = 100):
    ax.scatter([a,b],[f(a),f(b)])
    eps = 1.e-3
    # fig, ax = plt.subplots()
    ax.plot(X,f(X))
    numIter = 0
    err = eps + 1
    c = a
    while numIter < nMax and err > eps:
        numIter += 1
        a = c
        c = a-f(a) / fprime(a)
        err = abs(f(c))
        ax.scatter([c], [f(c)])
    ax.plot(X,f(a) + (X-a) * fprime(a))

NewtonWithPlot(f, fprime, a, b, eps, nMax = 4)
ax.grid()
```
