## 6.8 Positional vs. keyword arguments

Consider the following function:

```python
def f(x: int, y: int, z: int):
    'A simple function used to demonstrate calling sequences'
    print(f"x: {x}")
    print(f"y: {y}")
    print(f"z: {z}")
```

As we saw in the past, in the local scope of `f`, `x` take the value of the first argument, `y` that of the second, and `z` that of the third:

```python
f(1, 2, 3)
```

```python
f(3,2,1)
```

Consider however the following line of code:

```python
f(y = 1, x = 2, z = 3)
```

In `f(1, 2, 3)`, python matched the arguments with their value from their position. In `f(y = 1, x = 2, z = 3)`, python use *positional matching*. This statement explicitly says that the value 1 sould be associated to the parameter `y` etc 

The later is longer but more expressive when the function definition uses meaningful variable names. When keyword arguments are used, their order does not matter (*i.e.* `f(x = 1, y = 2, z = 3)` and `f(y = 2, x = 1, z = 3)` will produce the same result.

```python
f(x = 1, y = 2, z = 3)
```

```python
f(y = 2, x = 1, z = 3)
```

It is possible to combine positional and keyword arguments under the following condition:
* positional arguments precede keyword arguments (*i.e.* `f(1, 2, z = 3)` is OK but `f(1, y = 2, 3)` is not. This is because matching arguments can be ambiguous in this case.
* argument cannot be repeated. (*i.e.* `f(1, x=2, z=3)`) is not permitted.

```python
x = 1
y = 2
z = 3
f(x,y,z)
```

```python
f(x, y = z, z = y)
```

```python
f(x, z = y, y = z)
```

## 6.9 Default values

It is possible to assign a default value to an argument. In this case, it can be omitted in the calling sequence.


```python
def f(x: int, y: int, debug: bool = False) -> int:
    if debug:
        print(f'x: {x}, y: {y}')
    return x + y
```

```python
f(1,1)
```

```python
f(1,1,True)
```

When a function admits multiple optional argument and several are missing in the calling sequence, python matches arguments by position, unless of course keywords are given...

```python
def g(x: int, y: int, debug1: bool = False, debug2: bool = False) -> int:
    if debug1:
        print(f'debug1 is True, x: {x}, y: {y}')
    if debug2:
        print(f'debug2 is True, x+y: {x + y}')
    return x + y
g(1,1)
```

```python
g(1,1,True)
```

```python
g(1,1,True, True)
```

```python
g(1,1,debug2 = True)
```

# Dealing with an unknown number of input parameters

Sometimes we will want to write a function to perform a mathematical operation on an unknown quantity of input parameters. For example, maybe I want to write a function `average()` which takes as input an arbitrary number of integers and computes their average. One way of accomplishing this is by forcing the user to prepackage their data into a container object like a list or a tuple. We would obtain something like the following implementation:

```python
def average(nums : list[int]) -> float:
    """
    Takes as input a list of integers and returns
    their average (as a floating point number).

    Recall: The average of the numbers a1,...,an can be computed as 
    1/n(a1+...+an).
    """
    avg = 0.0
    for i in nums:
        avg += i
    avg = avg/len(nums)
    return avg
```

However, this feels like somewhat of a hack. It would be nice if Python allowed us to write functions which actually took an arbitrary number of parameters.

# *args command

Python's design philosophy revolves around making the language as flexible as possible. To that end, Python allows the user to **overload** function inputs with the *args command. This allows the user to provide as many input parameters as they like, all of which will be packed inside of a tuple with name followed after * (it is common to choose args as the name for this tuple).

```python
def func_with_args(f,*args):
    for i in args:
        print(f"f({i}) = {f(i)}.")

print(func_with_args(lambda x : x**2, 1,1.7,5,10,100))
```

### Revisiting our previous example - average()

We can use *args to write a nicer version of the average function we wrote previously.

```python
def average(*nums : int) -> float:
    '''
    Takes as input a collection of integers
    and returns their average.
    '''
    avg = 0.0
    for i in nums:
        avg += i
    avg = avg/len(nums)
    return avg
```

```python
print(average(1,2,3,10,100))
```

Combining *args with keyword arguments

```python
def average3(*args, verbose = False):
    if verbose:
        print(f'Computing the average of {len(args)} arguments in {args}')
    avg = 0
    for i in args:
        avg += i
    avg = avg / len(args)
    return avg
```

```python
print(average3(1,2,3, verbose = True))
```

On a similar note. Suppose we had the following function taking two postional arguments.

```python
import math
def norm2D(x: float,y: float) -> float:
    return math.sqrt(x**2+y**2)
```

This is a nice function. However, what if the user wanted to naturally input a tuple `a` of lenght 2 representing a vector. We could do the work to unpack this input from the user.

```python
a = (1,2)

print(norm2D(a[0],a[1]))

# or we could name the values of a
a1,a2 = a
print(norm2D(a1,a2))
```

However, this syntax is not ideal. Another way of doing this is the following:

```python
print(norm2D(*a))
```

We can combine both approaches to make an n-dimensional norm function.

```python
def normnD(*argv):
    norm = 0
    for i in argv:
        norm += i**2
    return math.sqrt(norm)
```

### Example 1 - using *args

Write a function which evaluates a polynomial of any degree at a value `a`, by accepting is coefficients as a sequence of *args values.  

```python
def poly_eval(a : float, *coeffs)-> float:
    eval = 0.0
    for i in range(len(coeffs)):
        eval += coeffs[i]*a**i
    return eval
```

### Example 2 - using *args

Write a function `compose()` which takes as input a floating point number `x` and a collection of functions and computes the result of evaluating the composition of the functions at `x`. For instance, `compose(x,f,g)` should return `f(g(x))` and `compose(x,f,g,h)` should return `f(g(h(x)))`. 

```python
def compose(x,*funs):
    eval = x
    for f in reversed(funs):
        eval = f(eval)
    return eval
```

# **kwargs 

The **kwargs command is the more structured cousin of the *args command. Instead of allowing the user to pass arbitrarily many *positional* input parameters, to be packed into a tuple, the **kwargs command allows the user to specify arbitrarily many *keyword* parameters, to be packed into a dictionary.

For a motivating example, consider the `dict()` constructor function:

```python
dict(a = 1, b = 2, c = 3)
```

This is easier (and less error prone) to write compared to this:

```python
{'a' : 1, 'b' : 2, 'c' : 3}
```

Consider the following function which allows us to print out $n$ evenly spaced evaluations of a function $f$ on an interval $[a,b]$.

```python
def tabulate(f, a, b, n):
    for i in range(n):
        x = a + (b-a) * i / (n-1) 
        print(f"f({x:.4f}) = {f(x):.4f}") # x:.4f means print x with 4 decimals only. We'll see that soon
```

```python
def g(x):
    return x**2
tabulate(lambda x : x**2, 0, 1, 3)
```

Recall now the "smooth absolute value" function we defined earlier in this section:

$$
    f_\varepsilon(x) = \sqrt{x^2+\varepsilon^2}.
$$

```python
import mathepsilon = 1e-
def smoothAbs(x: float, 1) -> float:
    return math.sqrt(x**2+epsilon**2)

# This will tabulate smoothAbs using the default value of epsilon
tabulate(smoothAbs, -0.5, 0.5, 11)
```

What if we want to tabulate `smoothAbs` for a different value of $\varepsilon$?

A simple modification to our `tabulate` function would allow us to make it work for any function which, like `smoothabs` requires an additional keyword arguments, *even if we don't know exactly what these arguments are*:

```python
def tabulate2(f, a, b, n, **kwargs):
    for i in range(n):
        x = a + (b-a) * i / (n-1) 
        print(f"f({x:.4f}) = {f(x,**kwargs):.4f}") # x:.2f means print x with 2 decimals only. We'll see that soon
```

```python
tabulate2(smoothAbs, -0.5, 0.5, 11, epsilon = 0.01)
```

```python
tabulate2(g, -0.5, 0.5, 11)
```

See also how we could use `tabulate2` with a function which takes ore than one keyword arguments:

```python
def linearFunction(x, slope = 0, offset = 0):
    return slope * x + offset
print(linearFunction(1, 2, 3))
```

```python
print(tabulate2(linearFunction,0,3,10,slope=2,offset=4))
```

```python
# another use of **kwargs
d = {'slope' : 1, 'offset' : 3}

print(linearFunction(1,**d))
```

## Example 1 - **kwargs 

Use kwargs to write a function called `circle()` which takes as input either the **radius**, **diameter**, or **circumference** of a circle and returns its area.

```python
import math
pi = math.pi

def area_of_circle(**kwargs):
    if 'radius' in kwargs:
        return pi*kwargs['radius']**2
    elif 'diameter' in kwargs:
        return pi*(kwargs['diameter']/2)**2
    elif 'circumference' in kwargs:
        return (kwargs['circumference']**2)/(4*pi)
    return 'Not enough information provided.' 

print(area_of_circle(radius = 2))
print(area_of_circle(diameter = 4))
print(area_of_circle(circumference = 4*pi))
print(area_of_circle(something_else = 5))
```
