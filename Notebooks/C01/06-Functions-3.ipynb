## 6.8 Positional vs. keyword arguments

Consider the following function:

```python
def f(x: int, y: int, z: int):
    'A simple function used to demonstrate calling sequences'
    print(f"x: {x}")
    print(f"y: {y}")
    print(f"z: {z}")
```

As we saw in the past, in the local scope of `f`, `x` take the value of the first argument, `y` that of the second, and `z` that of the third:

```python
f(1, 2, 3)
```

Consider however the following line of code:

```python
f(y = 1, x = 2, z = 3)
```

In `f(1, 2, 3)`, python matched the arguments with their value from their position. In `f(y = 1, x = 2, z = 3)`, python use *positional matching*. This statement explicitly says that the value 1 sould be associated to the parameter `y` etc 

The later is longer but more expressive when the function definition uses meaningful variable names. When keyword arguments are used, their order does not matter (*i.e.* `f(x = 1, y = 2, z = 3)` and `f(y = 2, x = 1, z = 3)` will produce the same result.

```python
f(x = 1, y = 2, z = 3)
```

```python
f(y = 2, x = 1, z = 3)
```

It is possible to combine positional and keyword arguments under the following condition:
* positional arguments precede keyword arguments (*i.e.* `f(1, 2, z = 3)` is OK but `f(1, y = 2, 3)` is not. This is because matching arguments can be ambiguous in this case.
* argument cannot be repeated. (*i.e.* `f(1, x=2, z=3)`) is not permitted.

```python
x = 1
y = 2
z = 3
f(x,y,z)
```

```python
f(x, y = z, z = y)
```

```python
f(x, z = y, y = z)
```

## 6.9 Default values

It is possible to assign a default value to an argument. In this case, it can be omitted in the calling sequence.


```python
def f(x: int, y: int, debug: bool = False) -> int:
    if debug:
        print(f'x: {x}, y: {y}')
    return x + y
```

```python
f(1,1)
```

```python
f(1,1,True)
```

When a function admits multiple optional argument and several are missing in the calling sequence, python matches arguments by position, unless of course keywords are given...

```python
def g(x: int, y: int, debug1: bool = False, debug2: bool = False) -> int:
    if debug1:
        print(f'debug1 is True, x: {x}, y: {y}')
    if debug2:
        print(f'debug2 is True, x+y: {x + y}')
    return x + y
g(1,1)
```

```python
g(1,1,True)
```

```python
g(1,1,True, True)
```

```python
g(1,1,debug2 = True)
```
