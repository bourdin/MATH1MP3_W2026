Now that we have discussed bisection search as an informal algorithm  we are ready to implement it in Python. 

Let's write a function called `bisection` which locates a root of a function $f(x)$ on a closed interval $[a,b]$ if it exists. Such a function needs four parameters: a function `f`; a left endpoint `a`; a right endpoint `b`; and a tolerance `tol`, that gives us the following end condition, if `abs(b-a)< tol` end the bisection algorithm.

```python
%matplotlib ipympl
import matplotlib.pyplot as plt

import numpy as np
```

```python
# we need this import statement to type annotate parameters
# of function type.
from collections.abc import Callable

def bisection(f : Callable[[float],float],a : float,b : float,tol : float) -> float | tuple[float,float] :
    '''
    An implementation of the bisection method for root finding.
    Takes as input a function `f`, an interval `[a,b]`, and a tolerance
    `tol`. Proceeds as follows:
    0. Let c = (a+b)/2 (the midpoint)
    1.  If f(a) = 0 return a
        If f(b) = 0 return b
        If f(c) = 0 return c
    2. Check if abs(b-a) < tol:
        If yes, return `(a,b)`.
        Otherwise...
    3. If f(a)*f(c)<0:
        b = c
        If f(c)*f(b)<0:
        a = c
    4. Repeat!
    '''
    if f(a)*f(b)< 0:
        while abs(a - b) >= tol:
            c = (a + b)/2 #midpoint of the interval
            if f(a) == 0.0 :
                return f"Found a root at x = {a}."
            elif f(b) == 0.0:
                return f"Found a root a x = {b}."
            elif f(c) == 0.0:
                return f"Found a root a x = {c}."
            elif f(a)*f(c) < 0:
                b = c
            else:
                a = c
        return f'Last interval checked ({a},{b})'
    else:
        return f" I don't know ask Newton"
```

# Question

In this lecture we identified two main drawbacks of bisection method for root finding. 
1. If multiple roots exist within an interval some can be missed.
2. If $f(x)$ has a repeated root on an interval, bisection method will not find it.

We can (1) by simply making our interval smaller. However, (2) is harder to fix. For this, the next best thing is called **Newton's method**.

```python
def fib(n: int) -> int:
    '''
    function that takes in an integer n
    and returns the nth fibonacci number
    '''
    i = 1 # ctr (going from 1 to n)
    fn = 1
    fn_plus_one = 1
    while i <= n:
        F = fn + fn_plus_one
        fn = fn_plus_one
        fn_plus_one = F
        i+=1
    return F
```
