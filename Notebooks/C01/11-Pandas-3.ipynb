```python
import pandas as pd
titanic = pd.read_csv("titanic.csv")
titanic.head()
```

```python
print('average passenger age: ', titanic.Age.mean())
print('average male passenger age: ', titanic[titanic.Sex == 'male'].Age.mean())
print('average female passenger age: ', titanic[titanic.Sex == 'female'].Age.mean())
```

```python
# Another way to do this:
# 1. select the Age and Sex columns:
#.  titanic[["Sex", "Age"]]
# 2. split into groups based on "Sex": 
# titanic[["Sex", "Age"]].groupby("Sex")
# 3. get the mean of each group
titanic[["Sex", "Age"]].groupby("Sex").mean()
```

```python
# Average age of survivors vs casualties:
titanic[["Survived", "Age"]].groupby("Survived").mean()
```

```python
# describe method prints some statistical info:
titanic[titanic.Survived == 1].describe()
```

```python
titanic[["Survived", "Age"]].groupby("Survived").describe()
```

```python
titanic.groupby("Sex").mean(numeric_only=True)
```
