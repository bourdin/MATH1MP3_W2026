### Pandas

Just as Numpy offers us the new data type of Numpy nd-arrays, Pandas comes with two new data types called **series** and **data frames**.

![Panda picture](KungFuPanda.png)

### Series

A **series** is a one-dimensional labeled array.

```python
a = [1, 7, 2]
ser1 = pd.Series(data = a)

print(ser1)
```

```python
print(ser1[2])
```

```python
ser2 = pd.Series(data = a, index = ["x", "y", "z"])
print(ser2)
```

```python
d = {"a" : 1, "b" : 2, "c" : 3}
ser = pd.Series(data = d, index = ["a","b","c"])
print(ser)
```

### Labels are helpful 

Suppose we have two datasets. They have some common data points, but they do not overlap completely. How can we make sure that arithmetic operations on them are well defined?

```python
hw1_data = {'John' : 72.0, 'Nathan' : 100.0, 'Matthew' : 88.9}
hw2_data = {'Nathan' : 100.0, 'Matthew' : 93.9}

hw1 = pd.Series(data = hw1_data)
hw2 = pd.Series(data = hw2_data)
```

```python
hw_avg = (hw1 + hw2)/2

hw_avg
```

**Notice:** The same task would a lot more work with Numpy arrays.

**However** this brings up an interesting point. Oftentimes our data will be incomplete. Pandas deals with incomplete data by labelling it with *NaN* standing for *Not a Number*. Dealing with missing data is one of the most important topics in data science (falling into the larger category of data sanitization). We will discuss how Pandas can help us deal with *NaN* values shortly.

### DataFrames

### Example calculation involving grades

```python
df = pd.read_csv('grades.csv')
```

### Final grading scheme
1. Scheme 1
- 10% Quizzes
- 20% HW
- 40% Exams (20% each)
- 30% Final Exam
2. Scheme 2
- 10% Quizzes
- 20% HW
- 20% Exams (Best 1 of 2)
- 50% Final Exam

```python
final_grade1 = 0.1*df[' Quizzes average'] + 0.2*((df[' HW1 mark']+ df[' HW2 mark']+ df[' HW3 mark'])/3) + 0.2*(df[' Exam1 mark']) + 0.2*(df[' Exam2 mark'])+0.3*(df[' Final exam mark'])

final_grade2 = 0.1*df[' Quizzes average'] + 0.2*((df[' HW1 mark']+ df[' HW2 mark']+ df[' HW3 mark'])/3) + 0.2*(df[[' Exam1 mark', ' Exam2 mark']].max(axis = 1))+0.5*(df[' Final exam mark'])

df.insert(10,'Final grade scheme 1', final_grade1)
df.insert(11, 'Final grade scheme 2', final_grade2)
```

```python
final_grade = df[['Final grade scheme 1', 'Final grade scheme 2']].max(axis = 1)
df.insert(12, 'Final grade', final_grade)
```

```python
df
```

### Dealing with missing data

Let's now try to accomplish the same goal as before, but this time with a more realistic data set.

```python
gradeBook = pd.read_csv('grades_missing.csv', na_values = [' '])
```

```python
gradeBook
```

We must choose how we deal with NaN values. In the case of grades, inserting zero works, but in other data sets this will not make sense. For instance, in some cases, inserting the mean of the data set is best.

```python
gradeBook = gradeBook.fillna(0)
```

```python
final_grade1 = 0.1*gradeBook[' Quizzes average'] + 0.2*((gradeBook[' HW1 mark']+ gradeBook[' HW2 mark']+ gradeBook[' HW3 mark'])/3) + 0.2*(gradeBook[' Exam1 mark']) + 0.2*(gradeBook[' Exam2 mark'])+0.3*(gradeBook[' Final exam mark'])

final_grade2 = 0.1*gradeBook[' Quizzes average'] + 0.2*((gradeBook[' HW1 mark']+ gradeBook[' HW2 mark']+ gradeBook[' HW3 mark'])/3) + 0.2*(gradeBook[[' Exam1 mark', ' Exam2 mark']].max(axis = 1))+0.5*(gradeBook[' Final exam mark'])

gradeBook.insert(10,'Final grade scheme 1', final_grade1)
gradeBook.insert(11, 'Final grade scheme 2', final_grade2)
```

```python
final_grade = gradeBook[['Final grade scheme 1', 'Final grade scheme 2']].max(axis = 1)
gradeBook.insert(12, 'Final grade', final_grade)
```

```python
gradeBook
```

Depending on the kind of data, sometimes it is best to remove all data that contains *NaN* values. For instance, when getting ethics approval for a research project involving a survey, it is sometimes required that all partially completed surveys be removed from the data set.

We can use the `dropna()` command for this.
