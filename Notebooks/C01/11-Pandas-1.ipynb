- We have seen how Numpy arrays are great tools for working with data of a uniform type.
    - In Pythonic words, we say Numpy arrays are homogeneous containers.
    - This allows Numpy arrays to have a single type - called their dtype.

```python
A = np.array([[1,2,3],[4,5,6]])
B = np.array([[1.,2.,3.],[4.,5,6]])
print(f"A = {A}")
print(f"B = {B}")
```

```python
print(f"A has dtype : {A.dtype}")
print(f"B has dtype : {B.dtype}")
```

However, we are oftentimes faced with data that is not homogeneous. In these cases, it would be nice to have a library with some of the similar functionality as Numpy i.e. **vectorization** and **columnization**, without the constraint of homogeneity. 

Such a library is **Pandas**.

### Pandas

Just as Numpy offers us the new data type of Numpy nd-arrays, Pandas comes with two new data types called **series** and **data frames**.

### Series

A **series** is a one-dimensional labeled array.

```python
a = [1, 7, 2]
ser1 = pd.Series(data = a)

print(ser1)
```

```python
print(ser1[2])
```

```python
ser2 = pd.Series(data = a, index = ["x", "y", "z"])
print(ser2)
```

```python
d = {"a" : 1, "b" : 2, "c" : 3}
ser = pd.Series(data = d, index = ["a","b","c"])
print(ser)
```

### DataFrames

A **DataFrame** is a two-dimensional labeled data structure that consists of columns and rows. A DataFrame represents a spreadsheet-like data structure where each column is a **Series**. 


```python
data = {
    "Name" : ["John", "Nathan", "Lily", "Corrin"],
    "Age" : [26,27,19,26],
    "City" : ["Hamilton", "Dundas", "Halifax", "Charlottetown"]
}

df = pd.DataFrame(data)

df
```

```python
data = [
    ["John", 26, "Hamilton"],
    ['Nathan', 27, "Dundas"],
    ["Lily", 19, "Halifax"],
    ["Corrin", 26, "Charlottetown"]
]

unique_id = [400130523, 400127887, 40130877, 555555555]

cols = ["Name", "Age", "City"]

df = pd.DataFrame(data, columns = cols, index = unique_id)

df
```

```python
df["Name"]
```

```python
df["City"]
```
