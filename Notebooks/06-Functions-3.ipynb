## 6.8 Positional vs. keyword arguments

Consider the following function:

````python
def f(x: int, y: int, z: int):
    'A simple function used to demonstrate calling sequences'
    print(f"x: {x}")
    print(f"y: {y}")
    print(f"z: {z}")
````

As we saw in the past, in the local scope of `f`, `x` take the value of the first argument, `y` that of the second, and `z` that of the third:

````python
f(1, 2, 3)
````

Consider however the following line of code:

````python
f(y = 1, x = 2, z = 3)
````

In `f(1, 2, 3)`, python matched the arguments with their value from their position. In `f(y = 1, x = 2, z = 3)`, python use *positional matching*. This statement explicitly says that the value 1 sould be associated to the parameter `y` etc 

The later is longer but more expressive when the function definition uses meaningful variable names. When keyword arguments are used, their order does not matter (*i.e.* `f(x = 1, y = 2, z = 3)` and `f(y = 2, x = 1, z = 3)` will produce the same result.

````python
f(x = 1, y = 2, z = 3)
````

````python
f(y = 2, x = 1, z = 3)
````

It is possible to combine positional and keyword arguments under the following condition:
* positional arguments precede keyword arguments (*i.e.* `f(1, 2, z = 3)` is OK but `f(1, y = 2, 3)` is not. This is because matching arguments can be ambiguous in this case.
* argument cannot be repeated. (*i.e.* `f(1, x=2, z=3)`) is not permitted.

````python
x = 1
y = 2
z = 3
f(x,y,z)
````

````python
f(x, y = z, z = y)
````

````python
f(x, z = y, y = z)
````

## 6.9 Default values

It is possible to assign a default value to an argument. In this case, it can be omitted in the calling sequence.


````python
def f(x: int, y: int, debug: bool = False) -> int:
    if debug:
        print(f'x: {x}, y: {y}')
    return x + y
````

````python
f(1,1)
````

````python
f(1,1,True)
````

When a function admits multiple optional argument and several are missing in the calling sequence, python matches arguments by position, unless of course keywords are given...

````python
def g(x: int, y: int, debug1: bool = False, debug2: bool = False) -> int:
    if debug1:
        print(f'debug1 is True, x: {x}, y: {y}')
    if debug2:
        print(f'debug2 is True, x+y: {x + y}')
    return x + y
g(1,1)
````

````python
g(1,1,True)
````

````python
g(1,1,True, True)
````

````python
g(1,1,debug2 = True)
````

## 6.10 Handling a variable number of arguments: unpack operators

Say that we want to write a function that computes the average of some numbers, but don't know *a-priori* how many. One solution would be to pass the numbers in a single list or tuple:   

````python
def average1(listOfNumbers: list[float]) -> float:
    ''' Compute the average of a bunch of numbers'''
    avg = 0
    for i in listOfNumbers:
        avg += i
    avg = avg / len(listOfNumbers)
    return avg
````

````python
print(average1([1]))
print(average1([1,2]))
print(average1([1,2,3]))
print(average1([1,2,3,4]))
````

Technically this works, but there is a more elegant way that does not require the additional step of putting the numbers into a list of a tuple, using the *unpack*  operator `*`: 

````python
def average2(*args):
    avg = 0
    for i in args:
        avg += i
    avg = avg / len(args)
    return avg

````

````python
print(average2(1))
print(average2(1, 2))
print(average2(1, 2, 3))
print(average2(1, 2, 3, 4))
````

The way this works is that all inside `average2`, `args` is a tuple containing the *positional* arguments to the function.

````python
def average3(*args, verbose = False):
    if verbose:
        print(f'Computing the average of {len(args)} arguments in {args}')
    avg = 0
    for i in args:
        avg += i
    avg = avg / len(args)
    return avg
````

````python
print(average3(1, verbose = False))
print(average3(1, 2, verbose = True))
print(average3(1, 2, 3))
print(average3(1, 2, 3, 4))
````

Conversely, suppose we have a function taking several positional arguments:

````python
import math
def norm2D(x: float,y: float) -> float:
    return math.sqrt(x**2+y**2)
````

````python
print(norm2D(3,4))
````

Say now that we have a variable `a` representing a vector in the plane as a list of length 2. We cannot call `norm2D(a)` since `norm2D` expects 2 arguments. Instead, we can *unpack* `a` in the call to `norm2D`:

````python
a = (3,4)
# norm2D(a) this is not valid

# this is valid but ugly!
print(norm2D(a[0], a[1]))
````

````python
# This is neat!
print(norm2D(*a)) 
````

And of course, we can combine both trick:

````python
def normnD(*argv):
    norm = 0
    for i in argv:
        norm += i**2
    return math.sqrt(norm)
````

````python
print(normnD(3,4))
a = [3,4]
# This wont work:
# print(normnD(a))
# This will:
print(normnD( *a ))
````

### Unpacking keyword arguments
The same approach works for keyword arguments, but it is less intuitive. Consider the following function:
```
def f(x, **kwargs):
    print (kwargs)
    return x
```

Inside `f`, `kwargs` is a dictionary containing any keyword argument that was passed in f.

````python
def f(x, **kwargs):
    print (f'kwargs is {kwargs}')
    return x
````

````python
f(0)
f(0, test = True)
````

We'll see a classical use for this feature when doing plots. Here is a somewhat convoluted example.
Let's start with a python function that approximates the derivative of a mathematical function:

````python
def tabulate(f, a, b, n):
    for i in range(n):
        x = a + (b-a) * i / (n-1) 
        print(f"f({x:.4f}) = {f(x):.4f}") # x:.4f means print x with 4 decimals only. We'll see that soon
````

````python
def g(x):
    return x**2
tabulate(g, 0, 1, 3)
````

Recall now the "smooth absolute value" function we defined earlier in this section:

$$
    f_\varepsilon(x) = \sqrt{x^2+\varepsilon^2}.
$$

````python
import math
def smoothAbs(x: float, epsilon = 1e-1) -> float:
    return math.sqrt(x**2+epsilon**2)

# This will tabulate smoothAbs using the default value of epsilon
tabulate(smoothAbs, -0.5, 0.5, 11)
````

What if we want to tabulate `smoothAbs` for a different value of $\varepsilon$?

A simple modification to our `tabulate` function would allow us to make it work for any function which, like `smoothabs` requires an additional keyword arguments, *even if we don't know exactly what these arguments are*:

````python
def tabulate2(f, a, b, n, **kwargs):
    for i in range(n):
        x = a + (b-a) * i / (n-1) 
        print(f"f({x:.4f}) = {f(x,**kwargs):.4f}") # x:.2f means print x with 2 decimals only. We'll see that soon
````

````python
tabulate2(smoothAbs, -0.5, 0.5, 11, epsilon = 0.01)
````

See how `tabulate2` still works with `g`

````python
tabulate2(g, -0.5, 0.5, 11)
````

See also how we could use `tabulate2` with a function which takes ore than one keyword arguments:

````python
def linearFunction(x, slope = 0, offset = 0):
    return slope * x + offset
print(linearFunction(1, 2, 3))
````

````python
print(linearFunction(1, offset = 3, slope = 2))
````

````python
tabulate2(linearFunction, -1, 1, 11, slope = 2, offset = 3)
````

### One last cool use of `**`
We can also think of `**` as an unpack operator, which convert a dictionary into a list of arguments:



````python
def f(x, **kwargs):
    print (f'kwargs is {kwargs}')
    return x

f(x, a = 1, b = 2, test = 'this is a test')
````

In the above, think of `**kwargs` as ` a = 1, b = 2, test = 'this is a test'` when `kwargs` is the dictionary `{'a': 1, 'b': 2, 'test': 'this is a test'}`.

Just like we used the `*` unpack operator to pass a list instead of multiple positional arguments, we can use the `**` unpack operator to pass a dictionary of arguments instead of a keyword arguments:

````python
d = {'offset': 0, 'slope': 2}
print(linearFunction(x,**d))
````
