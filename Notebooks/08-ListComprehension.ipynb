# Chapter 8: List comprehensions

List comprehension provide a simple way to create lists.

Say we want a list of the first n powers of 2. We can use a `for` loop.

````python
n = 5
L = []
for i in range(n):
    L.append(2**i)

print(f"L: {L}")
````

This has several flaws. 
   1. Although it is not obvious at first, growing a list is computationally intensive. It may require moving a lot of data in memory.
   1. This is 3 lines of code for something that can be described in a few words...

We could address 1 by creating L with the proper length directly:

````python
n = 5
L = [None,] * n
for i in range(n):
    L[i] = 2**i

print(f"L: {L}")
````

## 8.1 basic syntax
*List comprehensions* are a concise way to express the outcome of this loop. In this case, we could simply use:

````python
n = 5
L = [2**i for i in range(n)]
print(f"L: {L}")
````

The general syntax is 
```
newlist = [<expression> for <item> in <iterable>]
```

Example: a list of the first n multiples of 7

````python
L = [7*i for i in range(n)]
print(f"L: {L}")
````

````python
# first 15 even numbers:
print('first 15 even numbers: ',[2*i for i in range(15)])
# first 15 off numbers:
print('first 15 odd numbers: ',[2*i+1 for i in range(15)])
````

````python
# A list of 21 odd numbers starting with 13 (i.e 13 + 2*i with 0 <= i < 21)
odds = [13 + 2*i for i in range(21)]
print(f"odds: {odds}")
````

````python
# given a function f, print a table of values of f at 0, 1, ... n

def f(x: float) -> float:
    return x**2

print([f(x) for x in range(10)])
````

````python

# evaluate f at each point in a list

X = [1, 3, 4, 5, 6, 7, 2]
print(X)
print([f(x) for x in X])
````

````python
# a list of n values evenly distributed between a and b included 
a = 2
b = 3
n = 11

print([a + (b-a)/(n-1)*i for i in range(n)])
````

````python
# The coefficient of the derivative of a polynomial given its coefficients in increasing order 
P = [1,0,1]
Pprime = [p*(i+1) for i,p in enumerate(P[1:])]
print(Pprime)
````

## 8.2 List comprehensions with `if` statements

Say we want a list of numbers that are multiples of 7 but not 3, *i.e.* numbers of the form $7k$ where $k$ is not divisible by 3.  This translates directly into the following

````python
# just multiples of 7 
L = [7*k for k in range(10)]
print(L)

# just multiples of 7 
L3 = [7*k for k in range(10) if k%3 != 0]
print(L3)
````

### The general form of a list comprehension is really:
   ```
   [<statement> for <item> in <iterable> if <statement>]
   ```

````python
# a list of first 10 even numbers
l = []
for i in range(10):
    l = l + [2*(i+1)]
print(l)

l2 = []
for i in range(10):
    l2.append(2*(i+1))
print(l2)

l3  = [2*(i+1) for i in range(10)]
print(l3)

l4 = [i+1 for i in range(20) if (i+1)%2 == 0]
print(l4)

l5 = [i+1 for i in range(20) if i%2 != 0]
print(l5)
````

````python
# Given a list of grades, compute the alphabetical grade
# using the following simplified grading scale:
# 
#      grade < 1:  F
# 1 <= grade < 2: D
# 2 <= grade < 3: C
# 3 <= grade < 4: B
# 4 <= grade    : A
def alphaGrade(grade: float) -> str:
    if grade < 1:
        return "F"
    elif grade < 2:
        return "D"
    elif grade < 3:
        return "C"
    elif grade < 4:
        return "B"
    else:
        return "A"
    

# Generate fake grades: 20 numbers using a Gaussian distribution (bell curve) centered at 3  with stddev 2.5
import numpy as np
n = 20
grades = [min(max(0,np.random.normal(3., 2.5)),5) for i in range(n)]
# Note that in this case the first statement fors not depend on i


letterGrades = [alphaGrade(grade) for grade in grades]
for g,l in zip(grades, letterGrades):
    print(f'{g:.02f} -> {l}')
````

Note that the expression itself can be something complicated, involving conditionals, for instance

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
# get a list of all fruits whose name contain an "a":

afruits = [f for f in fruits if "a" in f]
print(afruits)
````

````python

bfruits = [f  for f in fruits]

# instead of f, we want 
# if "a" in f:
#     f
# else:
#     "***"
afruits = [f if "a" in f else "***"  for f in fruits]
print(afruits)
    
#bfruits = [f if not f.startswith("b") else "*"*len(f) for f in fruits]
print(bfruits)
````

````python
## example: if...else in a list comprehension 

listA=[num**2 for num in range(10) if num%2!=0]  ## describe this list in words
print(listA)

listB=[num+1 if num%2!=0 else num for num in range(1,11)]  ## describe this list in words
print(listB)
````
