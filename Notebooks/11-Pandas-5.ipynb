## python `datetime`
python has special data type to represent dates and times, in the appropriately named `datetime` module. It also provides way to do basic algebra with dates

```python
import datetime as dt
today = dt.date(year = 2026, month = 3, day = 19)
print(today)

now = dt.datetime(year = 2026, month = 3, day = 19)
print(now)

eoc = dt.date(2026,4,4)
print(f"only {eoc - today} until the end of the term")

print(f'We are {dt.datetime.now() - dt.datetime(2001,1,1)} into the 21st century!')
```

*relative time* (difference between two dates) are represented by `timedelta` (or `datetimedelta`):

```python
td = today - dt.date(2021, 12, 1)
print(td)
print(type(td))
print(td.days)
```

```python

now100 = (dt.datetime.today() + dt.timedelta(hours = 24*100))
print(type(now100))
print(now100)
print(f'100 days from now is {now100.strftime('%A %B %d, %Y')}')
```

## time series with pandas

```python
import pandas as pd
```

Let's get some air quality data from the pandas tutorial. Note that we can load teh file directly from a URL instead of downloading it first:

```python
url = 'https://raw.githubusercontent.com/pandas-dev/pandas/refs/heads/main/doc/data/air_quality_no2.csv'
aq = pd.read_csv(url)
```

```python
aq.head()
```

Let's rename `date.utc` to something that python likes better:


Right now, the "timestamp" Series consists of string, i.e. pandas has not realized that these are date / time data on which we could do operations. Let's fix this:

```python
print(type(aq['datetime'][0]))
aq["datetime"] = pd.to_datetime(aq["datetime"])
print(type(aq['datetime'][0]))
```

Note that we could also have instructed pandas to process dates upon reading the file:

```python
aq = pd.read_csv(url, parse_dates=[0])
type(aq.datetime[0])
```

Now we can do operations on the dates too:

```python
print('First measurement: ', aq['datetime'].min())
print('Last measurement:  ', aq['datetime'].max())
print('Survey length:     ', aq['datetime'].max() - aq['datetime'].min())
```
