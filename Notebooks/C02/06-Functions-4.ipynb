# Chapter 6: Function (part IV) - recursion

In mathematics, many formulas are *recursive*. Typically, the $n^{th}$ step of an algorithm or a sequence is defined in terms of the $(n-1)^{th}$.
A classical example is the factorial:

$$
n! = \begin{cases}
1 & \text{ if } n = 0,\\
(n-1)! & \text{ otherwise.}
\end {cases}
$$

Noticing that this gives us $n! = 1. 2. 3. \dots . (n-1) . n$, we can implement the factorial function using a for loop:

```python
def factorial1(n: int) -> int:
    ''' COmpute n! using a for loop'''
    fact = 1
    for i in range(1, n+1):
        fact *= i
    return fact
```

```python
for i in range(8):
    print(f'{i}! = {factorial1(i)}')
```

We can also leverage the recursive definition right away: 

```python
def factorial2(n):
    if n == 0:
        return 1
    else:
        fact = factorial2(n-1)
        return n * fact
## We could have just done return n * factorial2(n-1)
```

```python
for i in range(8):
    print(f'{i}! = {factorial2(i)}')
```

Note how `factorial2` is called insode itself. This can be visualized this way:

```python
def factorial2(n):
    if n == 0:
        return 1
    else:
        print(f'calling factorial2({n-1})')
        fact = factorial2(n-1)
        print(f'returning factorial2({n-1})')
        return n * fact
```

```python
factorial2(5)
```

Another example: finding an item in a list

```python
def findIt(i, L):
    print(f'looking for {i} in {L}')
    if len(L) == 0:
        print('not found')
        return False
    if L[0] == i:
        print('Found it!')
        return True
    else:
        return findIt(i, L[1:])
```

```python
findIt(1, (1,2,3))
```

```python
findIt(4, [1, -1, 2, 3, 5])
```

Recursive functions are controversial. Some love them because they can really simplify some algorithms. Others hate it because their complexity can grow really quickly. Consider the Fibonacci sequence:
$$ f_1 = f_2 = 1,\ f_{n+1} = f_n + f_{n-1}$$

```python
def FibonacciNR(n: int) -> float:
    ''' Compute the n-th term of the Fibonacci sequence'''
    if n == 1 or n == 2:
        return 1
    fpprev = 1
    fprev = 1
    fib = 0
    for i in range(2,n):
        fib = fprev + fpprev
        fpprev = fprev
        fprev = fib
    return fib
```

```python
for i in range(1, 21):
    print(f'{i}: {FibonacciNR(i)}')
```

```python
def FibonacciR(n: int) -> float:
    ''' Compute the n-th term of the Fibonacci sequence'''
    if n == 1 or n == 2:
        return 1
    else:
        return FibonacciR(n-1) + FibonacciR(n-2)
```

```python
for i in range(1, 11):
    print(f'{i}: {FibonacciR(i)}')
```

## Application: the quicksort algorithm

The quicksort algorithm (Tony Hoare, 1959) is a "divide-and-conquer" algorithm for sorting a list.
Given a list L, we pick a "pivot" value p in L and split L into three parts: 
* Ll: the numbers in L that are less than p
* Le: the numbers in L that are equal to p
* Lu: the numbers in L that are greater than p

We then apply quicksort to Ll and Lu and return 
Ll + Le + Lu

It is very easy to implement using recursivity

```python
def quicksort(L):
    if len(L) <= 1:
        return L
    p = L[len(L)//2]
    Ll = []
    Le = []
    Lu = []
    for l in L:
        if l < p:
            Ll.append(l)
        elif l == p:
            Le.append(l)
        else:
            Lu.append(l)
    print(f'pivot: {p}, {L} = {Ll} + {Le} + {Lu}')
    return quicksort(Ll) + Le + quicksort(Lu) 
```

```python
quicksort([1,3,6,-1,2,5,1])
```
