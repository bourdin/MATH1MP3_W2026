# Chapter 11: Pandas (continued) `DataFrame`

A `DataFrame` is a 2-dimensional labeled data structure with columns of potentially different types. One way to think about a `Dataframe` is as a two-dimensional list of data, where both rows and columns are structured, like a spreadsheet, or a `dict` of `Series`.
You can create a `DataFrame` from a list or dict of Series, a 2 axes ndarray

```python
import pandas as pd
import numpy as np
```

### Resume where we stopped in 11-1

```python
grades  = np.loadtxt('grades.csv',delimiter = ',', skiprows= 1, usecols=(3, 4, 5, 6, 7, 8, 9), max_rows=10)
keys = ("Quizzes_average", "HW1_mark", "HW2_mark", "HW3_mark", "Exam1_mark", "Exam2_mark", "Final_exam_mark")

gradesD = {}
for col, k in enumerate(keys):
    gradesD[k] = grades[:,col]

gradesL = []
for row in grades:
    record = {}
    for col, k in enumerate(keys):
        record[k] = row[col]
    gradesL.append(record)

rowD0 = pd.Series(gradesL[0])
rowD1 = pd.Series(gradesL[1])
rowD0.name = '620565656'
rowD1.name = '107856531'
print('gradesL[0]: ]', gradesL[0])
print('As panda series', rowD0)
```

```python
gradesDF = pd.DataFrame([rowD0, rowD1])
gradesDF
```

```python
print(gradesDF['HW1_mark'])
print(gradesDF.loc['620565656'])
print(type(gradesDF['HW1_mark']))
print(type(gradesDF.loc['620565656']))
```

A convenient shortcut: if the column names are valid python names (no spaces etc), it is possible to access columns directly using the "." operator:

```python
print(gradesDF['HW1_mark'])
print(gradesDF.HW2_mark)
```

Before we continue, let's generate a larger dataFrame by reading the entire grades file

```python
gradesDF2 = pd.read_csv("grades.csv", skipinitialspace=True, index_col=2)
gradesDF2.head() # same as gradesDF2[:5]
```

We now have the best of a list of dictionaries and a dictionary of lists:

```python
print(type(gradesDF2['HW3_mark']))
print(type(gradesDF2.loc[7943944]))
```

```python
# selecting rows:

# loc is label-based, iloc is index-based
print(gradesDF2.loc[114037001])
print(gradesDF2.iloc[2])
```

```python
# selecting columns
print(gradesDF2.HW1_mark[:10])
# or equivalently gradesDF2['HW1_mark'][:10]
```

```python
# Computing the HW1 average: 
print(gradesDF2.HW1_mark.describe())
```

```python
# Operation on columns
gradesDF2['HW_avg'] = (gradesDF2['HW1_mark'] + gradesDF2['HW2_mark'] + gradesDF2['HW3_mark']) / 3
gradesDF2.head()
```

## 11.3 Sorting

```python
gradesDF2.sort_values('HW_avg', ascending=False).head()
```

## 11.4 Selecting / filtering

### Selecting columns:

```python
gradesDF2[['HW1_mark', 'HW2_mark', 'HW3_mark']]
```

```python
# We could also have computed the grades average using:
gradesDF2[ ['HW1_mark', 'HW2_mark', 'HW3_mark']].mean(axis=1)
```

```python
# and the average for these items using
gradesDF2[ ['HW1_mark', 'HW2_mark', 'HW3_mark']].mean(axis=0)
```

### Selecting rows:

```python
A = gradesDF2['HW_avg'] >= 85
# print(A)

gradesDF2[A]
```

```python
B = (gradesDF2['HW_avg'] >= 80) & (gradesDF2['HW_avg'] < 85)
gradesDF2[B].sort_values('HW_avg', ascending=False)
```

```python
# short and very hard to read version:
gradesDF2[(gradesDF2['HW_avg'] >= 80) & (gradesDF2['HW_avg'] < 85)].sort_values('HW_avg', ascending=False)
```

```python
# Another way using a 'query'. Note the back quotes `
B2 = gradesDF2.query("80 <= `HW_avg` < 90")
```
