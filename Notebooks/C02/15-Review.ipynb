## A simple problem: local extrema of $\frac{sin x}{x}$ on $(-n \pi, n\pi)$

### Math background:
* **Critical points**: We say that $x_0$ is a critical point of a function $f$ if and only of $f'(x_0) = 0$ or $f$ is not diferentiable at $x_0$.
* **First derivative rule**: If a function $f$ is differentiable on an interval $(a,b)$ and admits a local extremum $x_0$ in $(a,b)$, then $f'(x_0) = 0$.
* **Second derivative rule**: If $f$ is twice differentiable in an interval containing $x_0$ and $f'(x_0) = 0$:
    * If $f''(x_0) > 0$ then $x_0$ is a *local minimizer* for $f$.
    * If $f''(x_0) < 0$ then $x_0$ is a *local maximizer* for $f$.
    * If $f''(x_0) = 0$, the test is inconclusive.
* **Mean value theorem**: If a function $f$ is continuous on a closed interval $[a,b]$ and differentiable on $(a,b)$, then there exists $c \in [a,b]$ such that $f'(c) = \frac{f(b) - f(a)}{b-a}$.
* **Corollary**: under the same asumptions, of $f(a) = f(b) = 0$, there exists at least one $c \in [a,b]$ such that $f'(c) = 0$.

#### Warning: 
numpy and suympy have different definitions of `sinc`: numpy uses $\frac{sin \pi x}{\pi x}$ while sympy uses the more common $\frac{\sin x}{x}$.


```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
import scipy
%matplotlib widget
```

```python
def f(x):
    return np.sinc(x / np.pi)

x = sp.Symbol('x')
ff = sp.sinc(x)
```

```python
### Let's plot f with numpy / matplotlib
plt.close('all')
n = 5
fig, ax = plt.subplots()
X = np.linspace(-n*np.pi, n*np.pi, 201)
ax.plot(X,f(X), label = 'sinc')
# ax.plot(X,np.sin(X), label = 'sin')
ax.spines[['left', 'bottom']].set_position('zero')
ax.spines[['top', 'right']].set_visible(False)
ax.legend(loc='upper left')
ax.set_xlabel('x')
ax.set_xlabel('y')
ax.set_title('Extrema of the cardinal sine')

# rootsoff = [ k*np.pi for k in range(-n, n+1) if k != 0 ]
# ax.scatter(rootsoff, [f(r) for r in rootsoff], label = 'k.pi')
```

Let's find the critical points of $f$. First, we need $f'$:

```python
ffprime = sp.simplify(sp.diff(ff,x))
fprime = sp.lambdify(x, ffprime)
ffprime
```

Let's now use `scipy.optimize.root_scalar` to find the critical points of $f$. We use initial guesses of the form $k\pi$, $k \in \mathbb{Z}$

```python
# Finding roots of f' take 1:
# Does not work well because Newton's method 
# gives little control over the root it converges to 
roots = []
for i in range(-n, n+1):
    r = scipy.optimize.root_scalar(fprime, x0 = i * np.pi)
    if r.converged:
        roots.append(r.root)
    else:
        print(f'Failed to find a root near {n}pi')
print(roots)
```

```python
plt.close('all')
n = 5
fig, ax = plt.subplots()
X = np.linspace(-n*np.pi, n*np.pi, 201)
ax.plot(X,f(X), label = 'sinc')
# ax.plot(X,np.sin(X), label = 'sin')
ax.spines[['left', 'bottom']].set_position('zero')
ax.spines[['top', 'right']].set_visible(False)
ax.legend(loc='upper left')
ax.set_xlabel('x')
ax.set_xlabel('y')
ax.set_title('Extrema of the cardinal sine')
ax.scatter(roots, [f(r) for r in roots])
```

```python
# Finding roots of f' take 2:
# use bisection method
# gives little control over the root it converges to 
roots = []
rootsoff = [k * np.pi for k in range(-n,n+1) if k != 0]
for i in range(len(rootsoff)-1):
    a = rootsoff[i]
    b = rootsoff[i+1]
    r = scipy.optimize.root_scalar(fprime, method = 'bisect', bracket = (a, b))
    if r.converged:
        roots.append(r.root)
    else:
        print(f'Failed to find a root near {n}pi, {r}')
print(roots)
# ax.scatter(roots, [f(r) for r in roots])
# fig.show()
```

```python
plt.close('all')
n = 5
fig, ax = plt.subplots()
X = np.linspace(-n*np.pi, n*np.pi, 201)
ax.plot(X,f(X), label = 'sinc')
# ax.plot(X,np.sin(X), label = 'sin')
ax.spines[['left', 'bottom']].set_position('zero')
ax.spines[['top', 'right']].set_visible(False)
ax.legend(loc='upper left')
ax.set_xlabel('x')
ax.set_xlabel('y')
ax.set_title('Extrema of the cardinal sine')
ax.scatter(roots, [f(r) for r in roots])
```

This almost worked, but could not find the extremum at 0. This is because $f'$ needs to be extended by continuity at 0 in order to define it.

```python
def fprime2(x):
    if x != 0:
        return fprime(x)
    else:
        return 0
```

```python
# Finding roots of f' take 2:
# use bisection method
# gives little control over the root it converges to 
roots = []
for i in range(len(rootsoff)-1):
    a = rootsoff[i]
    b = rootsoff[i+1]
    r = scipy.optimize.root_scalar(fprime2, method = 'bisect', bracket = (a, b))
    if r.converged:
        roots.append(r.root)
    else:
        print(f'Failed to find a root near {n}pi, {r}')
print(roots)

plt.close('all')
n = 5
fig, ax = plt.subplots()
X = np.linspace(-n*np.pi, n*np.pi, 201)
ax.plot(X,f(X), label = 'sinc')
# ax.plot(X,np.sin(X), label = 'sin')
ax.spines[['left', 'bottom']].set_position('zero')
ax.spines[['top', 'right']].set_visible(False)
ax.legend(loc='upper left')
ax.set_xlabel('x')
ax.set_xlabel('y')
ax.set_title('Extrema of the cardinal sine')

rootsoff = [ k*np.pi for k in range(-n, n+1) if k != 0 ]
ax.scatter(rootsoff, [f(r) for r in rootsoff], label = 'k.pi')
ax.scatter(roots, [f(r) for r in roots])
```

Let's finally use the second derivative rule to distinguish local minima from local maxima. We should also try to deal with 0 separately

```python
ffpprime = sp.diff(ff,x,2)
fpprime2 = sp.lambdify(x, ffpprime)
localmin = []
localmax = []
for r in roots:
    if r != 0.:
        # der2test = ffprime.subs(x, r).evalf()
        der2test = fpprime2(r)
        if der2test < 0.:
            localmax.append(r)
        elif der2test > 0.:
            localmin.append(r)
print(localmin)
print(localmax)
```

```python
plt.close('all')
n = 5
fig, ax = plt.subplots()
X = np.linspace(-n*np.pi, n*np.pi, 201)
ax.plot(X,f(X), label = 'sinc')
# ax.plot(X,np.sin(X), label = 'sin')
ax.spines[['left', 'bottom']].set_position('zero')
ax.spines[['top', 'right']].set_visible(False)
ax.legend(loc='upper left')
ax.set_xlabel('x')
ax.set_xlabel('y')
ax.set_title('Extrema of the cardinal sine')

rootsoff = [ k*np.pi for k in range(-n, n+1) if k != 0 ]
ax.scatter(rootsoff, [f(r) for r in rootsoff], label = 'k.pi')
ax.scatter(localmin, [f(r) for r in localmin], color = 'red')
ax.scatter(localmax, [f(r) for r in localmax], color = 'green')
```

## A boundary value problem
A classical problem consists in finding a function $u$ satisfying
$$
\begin{cases}
-u''(x) = f & \text{ in } (0,1),\\
u(0) = a,  & \\
u(1) = b,
\end{cases}
$$
where $a, b$ are two given numbers and $f$ is a given function. This problem appear in fluid and solid mechanics, in heat transfer, data analysis, and many other fields.

We are goint to try to approximate solutions of this problem using python in two different ways by approximating the values of the folution function $u$ at $N$ equidistributed points in $(0,1)$, which we will represent by a numpy array `U`.

### Explicit finite difference:
The solution array can be approximated through the following iterative algorithm:
Let $dx = 1/(N-1)$ and $0 < dt < dx^2/2$. 
* Initialize `U` such that `U_ = a` and `U_{N-1} = b`
* Iterate until convergence
     - Compute $dU$ such that $dU_i = -(U_{i-1} - 2dU_i + dU_{i+1}) / dx^2 - F_i$, $i = 1 \dots N-2$.
     - Compute $U \leftarrow U - dt dU$



```python
N = 101
u = np.zeros(N)
f = np.ones(N)
dx = 1/(N-1)
dt = dx**2/2 * 0.99

niter = 10000
du = np.zeros(N)
for i in range(niter):
    for i in range(1,N-1):
        du[i] = -(u[i-1] - 2*u[i] +u[i+1]) / dx**2 - f[i]
    u -= dt * du

plt.close('all')
fig, ax = plt.subplots()
ax.plot(u)
```

### Implicit finite differences:
`U` can also be computed directly as the solution of the linear system `KU = F`, where $K$ is a $N\times N$ matrix such that 
$$K[i,j] = \begin{cases}
\frac{-1}{dx^2} & \text{ if } 0 < i < N-1 \text{ and } j = i-1,\\
\frac{2}{dx^2} & \text{ if } 0 < i < N-1 \text{ and } j = i,\\
\frac{-1}{dx^2} & \text{ if } 0 < i < N-1 \text{ and } j = i+1,\\
1 & \text{ if } i = 0 \text{ and } j = 0,\\
1 & \text{ if } i = N-1 \text{ and } j = N-1,\\
0 & \text{ otherwise.}
\end{cases}
$$
and $F$ is a vector of length $N$ such that
$$
F_i = \begin{cases}
a & \text{ if } i = 0,\\
b & \text{ if } i = N-1,\\
f(i/(N-1)) & \text{ otherwise.}\\
\end{cases}
$$


```python
K = np.zeros((N,N))
F = np.ones(N)
K[0,0] = 1
K[N-1,N-1] = 1
F[0] = 0
F[N-1] = 0
for i in range(1,N-1):
    K[i,i-1] = -1. / dx**2
    K[i,i]   = 2. / dx**2
    K[i,i+1] = -1. / dx**2
```

```python
U = np.linalg.solve(K,F)
plt.close('all')
fig, ax = plt.subplots()
ax.plot(U)
```
