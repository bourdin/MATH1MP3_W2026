## 9.3 stacking, reshaping
Say that we have two np.arrays of identical length with one axis each and want to combine them
There are really three possible ways to do that: `vstack`, `hstack`, `dstack`.

```python
import numpy as np
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[-4, -5, -6], [-7, -8, -9]])
print(A)
print(B)
```

```python
print(np.vstack((A, B)))
print(np.hstack((A, B)))
print(np.dstack((A, B)))
```

Another operation is to *reshape* an array, for instance:

```python
A = [[1, 2, 3], [4, 5, 6]]
print(A)
print(np.reshape(A, (6)))
print(np.reshape(A, (6, 1)))
print(np.reshape(A, (1, 6)))
print(np.reshape(A, (3, 2)))
```

## 9.4 Example: managing grades

Let's pretend that we have stored student grades in two different nparray, one for assignments and the other for exams.
`assignments` consists of $n$ rows, each with the student ID, and 5 assignment marks
`exams` $n$ rows, each with the student ID, and 3 assignment marks.

* we can merge these two tables into a full grade list using `hstack`
* We could add records using `vstack`


```python
# We'll explain how we read this from a file later
assignments  = np.loadtxt('grades.csv',delimiter = ',', skiprows= 1, usecols=(2, 4, 5, 6), max_rows=10)
exams        = np.loadtxt('grades.csv',delimiter = ',', skiprows= 1, usecols=(2, 7, 8, 9), max_rows=10)
print(assignments)
print(exams)
```

```python
# Merging the arrays
allGrades = np.hstack((assignments, exams[:,1:]))
print(allGrades[:5,:])
```

```python
# Adding the records for a new student
newRow = [123456789, 85, 90, 95, 80, 85, 80]
allGrades = np.vstack( (allGrades, newRow) )
for row in allGrades:
    print(f'ID: {int(row[0])}, grades: {row[1:]}')
```

```python
help(np.average)
```

```python
# Adding a new column with the final grade given as
# F = (A1 + A2 + A3) / 3 * 0.25 + (E1 + E2 + E3) / 3 * .75)
#finalGrade = np.array([ np.average(row[1:4]) * .25 + np.average(row[4:7]) * .75 for row in allGrades])
finalGrade = np.array([np.average(row[1:4]) * .25 + np.average(row[4:7]) * .75 for row in allGrades])
print(finalGrade)
# allGrades2 = np.hstack( (allGrades, finalGrade) )


# finalGrades is a 'row' vector, so we need to reshape it iw we want to add it to our list of grades
numGrades = len(finalGrade)
print(numGrades)
print(np.reshape(finalGrade, (numGrades,1) ))
allGrades2 = np.hstack( (allGrades, np.reshape(finalGrade, (numGrades,1) )))
for row in allGrades2:
    print(f'ID: {int(row[0])}, grades: {row[1:]}')
```

```python
classAverages = np.average(allGrades2[:, 1:], (0))
print(classAverages)
print(classAverages[-1], np.average(classAverages[:3]) * .25 + np.average(classAverages[3:6])*.75)
```

## 9.7 Iterating over np.arrays

Consider a np.array of unknown shape. How do we cycle through all entries? The total number of nested foor loops will depend on the shape of the array.

```python
import numpy as np
A1 = np.array([1, 2, 3])
A2 = np.array([[1, 2, 3], [4, 5, 6]])
A3 = np.array([[[1, 2, 3], [4, 5, 6]],
              [[-1, -2, -3], [-4, -5, -6]],
              [[10, 20, 30], [40, 50, 60]]])
print(A1)
print(A2)
print(A3)

```

```python
maxA1 = A1[0]
for a in A1:
    if a > maxA1:
        maxA1 = a
        
print('A1: ', A1)
print('maxA1: ', maxA1)
```

```python
maxA2 = A2[0]
for a in A2:
    print(a)
    # if a > maxA2:
    #     maxA2 = a
        
print('A2: ', A2)
# print('maxA2: ', maxA2)
```

```python
maxA2 = A2[0,0]
for row in A2:
    for a in row:
        if a > maxA2:
            maxA2 = a
print('A2: ', A2)
print('maxA2: ', maxA2)
```

```python
for face in A3:
    for row in face:
        for a in row:
            print('A3: ', a)
```

numpy provides a nd iterator which will work in all cases:

```python
for a in np.nditer(A1):
    print('a: ', a)
for a in A1:
    print('a: ', a)
```

```python
for a in np.nditer(A2):
    print('a: ', a)

# same as 
for row in A2:
    for a in row:
        print('a: ', a)
```

```python
for a in np.nditer(A3):
    print('A3: ', a)
```

```python
def ndMax(A) -> float:
    maxA = None
    for a in np.nditer(A):
        if maxA == None or a > maxA:
            maxA = a
    return maxA

print(ndMax(A1))
print(ndMax(A2))
print(ndMax(A3))
```

```python
# Example: compute the L^p norm of a np.array:
def Lpnorm(A, p) -> float:
    norm = 0
    for a in np.nditer(A):
        norm += a**p
    return np.sqrt(norm)

print(Lpnorm(A1, 2))
print(Lpnorm(A2, 2))
print(Lpnorm(A3, 2))
```

Similarly, there is a `np.ndenumerate`

```python
for i,a in enumerate(A1):
    print(f'A1[{i}]  {a} ')

for i,row in enumerate(A2):
    for j, a in enumerate(row):
        print(f'A2[{i},{j}] = {a} ')
```

```python
for i, a in np.ndenumerate(A1):
    print(f'A1[{i}]  {a} ')

for i, a in np.ndenumerate(A2):
    print(f'A2[{i}]  {a} ')

for i, a in np.ndenumerate(A3):
    print(f'A3[{i}]  {a} ')

```

```python
def ndSum(A, B):
    if np.shape(A) != np.shape(B):
        print('Non matching shapes', np.shape(A), np.shape(B))
        return None
    C = np.empty(np.shape(A))
    # C_i = A_i + B_i
    for i, a in np.ndenumerate( A ):
        C[i] = a + B[i]
    return C

B1 = np.array([4, 5, 6])
B2 = np.eye(2,3 )
print(ndSum(A1, B1))
print(A2)
print(B2)
print(ndSum(A2, B2))
```

```python
for i in np.ndindex(np.shape(A2)):
    print(i)

# Better version of ndSum
def ndSum2(A, B):
    if np.shape(A) != np.shape(B):
        print('Non matching shapes', np.shape(A), np.shape(B))
        return None
    C = np.empty(np.shape(A))
    # C_i = A_i + B_i
    for i in np.ndindex( np.shape(A) ):
        C[i] = A[i] + B[i]
    return C

B1 = np.array([4, 5, 6])
B2 = np.eye(2,3 )
print(ndSum2(A1, B1))
print(A2)
print(B2)
print(ndSum2(A2, B2))
```

### Application: applying a function to a ndarray
We saw that numpy tries to do all operations component-wise. Given the 3d array `A3` and a function `f`, we can apply `f` to each term of `A3`: 

```python
def f(x: float) -> float:
    return 2*x
print('f(A1):\n',f(A1))
print('f(A2):\n',f(A2))
print('f(A3):\n',f(A3))
```

Sometimes, python needs help, though...

```python
def g(x:float)-> float:
    if x < 0:
        return -x
    else: 
        return x
print(g(1))
print(g(-1))
print(A1 < 0)
```

What happens here is that python substitute `A3` for `x` in `g`, and the statement `if A3 < 0:` is ambiguous.
We could create a *vectorized* version of `g`, but unless we are extra careful, it is no longer going to work for floats...

```python
def gVect(x):
    res = np.empty(x.shape) # same as np.shape(x)
    for i, val in np.ndenumerate(x):
        res[i] = g(val)
    return res
print('gVec(A1):\n',gVect(A1))
print('gVec(A1):\n',gVect(A2))
print('gVec(A1):\n',gVect(A3))

print('gVect(-1):', gVect(-1))
```

```python
gVect2 = np.vectorize(g)
print('gVec2(A1):\n',gVect2(A1))
print('gVec2(A2):\n',gVect2(A2))
print('gVec2(A3):\n',gVect2(A3))

print('gVect2(-1):', gVect2(-1))
```
